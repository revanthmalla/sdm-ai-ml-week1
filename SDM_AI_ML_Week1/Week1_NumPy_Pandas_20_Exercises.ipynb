## Part A: NumPy Exercises (1–10)

NumPy is used for working with numerical arrays and performing calculations efficiently.

### 1. Create a NumPy array

**Problem:** Create an array containing the numbers 10, 20, 30, 40, and 50.

```python
numbers = np.array([10, 20, 30, 40, 50])
print(numbers)
```

### 2. Create an array of zeros

**Problem:** Create a one-dimensional array containing five zeros.

```python
zeros = np.zeros(5, dtype=int)
print(zeros)
```

### 3. Create an array of ones

**Problem:** Create a one-dimensional array containing four ones.

```python
ones = np.ones(4, dtype=int)
print(ones)
```

### 4. Generate a sequence

**Problem:** Generate integers from 1 up to (but not including) 11.

```python
sequence = np.arange(1, 11)
print(sequence)
```

### 5. Create an array of even numbers

**Problem:** Create the even numbers from 2 to 20.

```python
even_numbers = np.arange(2, 21, 2)
print(even_numbers)
```

### 6. Reshape an array

**Problem:** Create numbers from 1 to 6 and reshape them into 2 rows and 3 columns.

```python
numbers = np.arange(1, 7)
matrix = numbers.reshape(2, 3)
print(matrix)
```

### 7. Access an array element

**Problem:** Create an array and print its third element (using index 2).

```python
values = np.array([5, 10, 15, 20, 25])
print("Third element:", values[2])
```

### 8. Slice an array

**Problem:** Print the first three elements of an array.

```python
values = np.array([10, 20, 30, 40, 50])
print(values[:3])
```

### 9. Perform arithmetic on an array

**Problem:** Add 5 to every element in the array.

```python
marks = np.array([60, 70, 80, 90])
updated_marks = marks + 5
print(updated_marks)
```

### 10. Calculate basic statistics

**Problem:** Find the minimum, maximum, and average of the given values.

```python
scores = np.array([45, 60, 75, 90, 80])
print("Minimum:", np.min(scores))
print("Maximum:", np.max(scores))
print("Average:", np.mean(scores))
```

## Part B: Pandas Exercises (11–20)

Pandas is used to organize, clean, filter, and analyse data in tables called DataFrames.

```python
# Sample student dataset used in the Pandas exercises
students = pd.DataFrame({
    "Name": ["Anu", "Bala", "Charan", "Divya", "Esha"],
    "Age": [19, 20, 19, 21, 20],
    "Department": ["CSE", "ECE", "CSE", "EEE", "ECE"],
    "Marks": [85, 72, 90, 68, 88]
})

students
```

### 11. Create a Pandas Series

**Problem:** Create a Series containing three subject marks.

```python
marks = pd.Series([85, 90, 78], index=["Maths", "Python", "Java"])
print(marks)
```

### 12. Create a DataFrame

**Problem:** Create a DataFrame with two columns: Name and Marks.

```python
data = {
    "Name": ["Asha", "Ravi", "Maya"],
    "Marks": [80, 75, 92]
}
df = pd.DataFrame(data)
print(df)
```

### 13. Display the first rows

**Problem:** Display the first three rows of the sample student dataset.

```python
print(students.head(3))
```

### 14. Find the shape of a DataFrame

**Problem:** Print the number of rows and columns in the student dataset.

```python
print("Rows and columns:", students.shape)
```

### 15. Select one column

**Problem:** Display only the Name column.

```python
print(students["Name"])
```

### 16. Select a row

**Problem:** Display the first row using its row position.

```python
print(students.iloc[0])
```

### 17. Filter rows

**Problem:** Display students who scored more than 80 marks.

```python
high_scorers = students[students["Marks"] > 80]
print(high_scorers)
```

### 18. Add a new column

**Problem:** Add a Result column. Students with marks of 75 or more pass; others need improvement.

```python
students_copy = students.copy()
students_copy["Result"] = np.where(students_copy["Marks"] >= 75, "Pass", "Needs Improvement")
print(students_copy)
```

### 19. Sort the DataFrame

**Problem:** Sort students by Marks from highest to lowest.

```python
sorted_students = students.sort_values(by="Marks", ascending=False)
print(sorted_students)
```

### 20. Group and calculate an average

**Problem:** Group students by Department and calculate the average marks for each department.

```python
department_average = students.groupby("Department")["Marks"].mean()
print(department_average)
```
